```python
import pandas as pd
import seaborn as sb
import numpy as np

from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split as tts
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

```python
df = pd.read_csv('Datasets\\Churn_Modelling.csv')
df.head()
```

```python
lb = LabelEncoder()
df['Geography'] = lb.fit_transform(df['Geography'])
df['Gender'] = lb.fit_transform(df['Gender'])

df
```

```python
df.info()
```

```python
df = df.drop(['RowNumber'],axis=1)
df = df.drop(['CustomerId'],axis=1)
df = df.drop(['Surname'],axis=1)
```

```python
y = df.pop('Exited')
x = df
```

```python
y.shape
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test = tts(x,y,test_size=.2,random_state=0)
```

```python
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

```python
x_train
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import LeakyReLU,PReLU,ELU
from keras.layers import Dropout
```

```python
classifier = Sequential()
```

```python
classifier.add(
    Dense(units=6,kernel_initializer='he_uniform',activation='relu',input_dim=10)
)
```

```python
classifier.add(
    Dense(units=6,kernel_initializer='he_uniform',activation='relu')
)
```

```python
classifier.add(
    Dense(units=1,kernel_initializer='glorot_uniform',activation='sigmoid')
)
```

```python
classifier.compile(
    optimizer='Adamax',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model_history = classifier.fit(x_train,y_train,batch_size=10,epochs=100)
```

```python
classifier.summary()
```

```python
y_pred = classifier.predict(x_test)
y_pred
```

```python
y_pred = (y_pred>.5)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
confusion_matrix(y_test,y_pred)[0]
```

```python
accuracy_score(y_pred,y_test)
```

```python
str(round(accuracy_score(y_test,y_pred),4)*100) + '%'
```
