```python
import cv2
from matplotlib import pyplot as plt
import numpy as np

def imshow(img):
    plt.imshow(img)
    plt.axis("off")
    plt.show()
```

```python
image = cv2.imread('color.jpg')

imshow(image)
image.shape
```

```python
gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
imshow(gray_image)
gray_image.shape
```

```python
hsv_image = cv2.cvtColor(image,cv2.COLOR_BGR2HSV)
imshow(hsv_image)
hsv_image.shape
```

```python
canny_image = cv2.Canny(gray_image,150,200)
imshow(canny_image)
```

```python
kernel = np.ones((1,1),np.uint8)
eroded_image = cv2.dilate(canny_image,kernel,iterations=1)
imshow(eroded_image)
```

```python
kernel1 = np.ones((5,5),np.uint8)
dilate_image = cv2.dilate(canny_image,kernel1,iterations=1)
imshow(dilate_image)
```

```python
hori_display = np.concatenate((canny_image,eroded_image,dilate_image),axis=1)
imshow(hori_display)
```

```python
image = cv2.imread('lion.jpg')
imshow(image)
```

```python
denoise_image= cv2.fastNlMeansDenoisingColored(
    image,None,20,20,7,15
)
imshow(denoise_image)
```

```python
hori_display = np.concatenate((image,denoise_image),axis=1)
imshow(hori_display)
```
