```python
from diyanLib.imports.cv import *
```

```python
def img():
    return np.zeros(shape=(512,512,3),dtype=np.int16)
```

```python
emptyImg = img()
plt.imshow(emptyImg)
```

```python
cv2.rectangle(
    emptyImg,
    pt1=(380,10),
    pt2=(500,150),
    color=(255,0,0),
    thickness=1)

plt.imshow(emptyImg)
```

```python
cv2.circle(emptyImg,
           center=(400,400),
           radius=100,
           color=(255,0,0),
           thickness=5
           )
plt.imshow(emptyImg)
```

```python
cv2.line(
    emptyImg,
    pt1=(0,0),
    pt2=(512,512),
    color=(255,0,0),
    thickness=5
)

plt.imshow(emptyImg)
```

```python
cv2.line(
    emptyImg,
    pt1=(0,512),
    pt2=(512,0),
    color=(255,0,0),
    thickness=5
)

plt.imshow(emptyImg)
```

```python
cv2.circle(emptyImg,
           center=(100,100),
           radius=100,
           color=(255,0,0),
           thickness=5
           )
plt.imshow(emptyImg)
```

```python
cv2.line(emptyImg,
         pt1=(0,256),
         pt2=(512,256),
         color=(255,0,0),
         thickness=5)

plt.imshow(emptyImg)
```

```python
emptyImg2 = img()
plt.imshow(emptyImg2)
```

```python
emptyImg2 = img()
emptyImg2 = emptyImg2.astype(np.uint8)

cv2.putText(
    emptyImg2,
    text='-'*9,
    org=(20,250),
    fontFace=cv2.FONT_HERSHEY_SIMPLEX,
    fontScale=4,
    color=(100,0,150),
    thickness=3,
    lineType=cv2.LINE_AA
)

plt.imshow(emptyImg2)
```
