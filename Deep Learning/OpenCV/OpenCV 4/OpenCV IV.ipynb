```python
from diyanLib.imports.cv import *
```

**Reading and displaying Image**

```python
img = cv2.imread('shapes.png')
cv2_imshow(img)
```

**Grayscale Image**

```python
gray_img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2_imshow(img)
```

```python
ret,thresh = cv2.threshold(gray_img,50,255,1)
cv2_imshow(thresh)
```

```python
contours,h = cv2.findContours(thresh,1,2)
```

```python
for cnt in contours:
    approx = cv2.approxPolyDP(cnt,.01*cv2.arcLength(cnt,True),True)
    n = len(approx)

    if n==6:
        print('Found Hexagon')
        cv2.drawContours(img,[cnt],0,255,10)

    elif n==3:
        print('Found triangle')
        cv2.drawContours(img,[cnt],0,255,10)

    elif n>9:
        print('Found Circle')
        cv2.drawContours(img,[cnt],0,255,10)

    elif n==4:
        print('Found Quadrilateral')
        cv2.drawContours(img,[cnt],0,255,10)

cv2_imshow(img)
```
