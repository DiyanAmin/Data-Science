```python
from diyanLib.imports.cv import *
```

```python
face_cascade = cv2.CascadeClassifier(f'Algorithms\\smile.xml')
```

```python
img = cv2.imread('images\\smiles.jpg')
cv2_imshow(img)
```

```python
gray_img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2_imshow(gray_img)
```

```python
face = face_cascade.detectMultiScale(gray_img,1.3,5)
print(face)
```

```python
for (x,y,w,h) in face:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),3)

cv2_imshow(img)
```
