```python
import pandas as pd
import numpy as np
from keras.models import Sequential
from keras.layers import Dense
```

```python
df = pd.read_csv('housing_data.csv',sep=',')
df.head()
```

```python
y = df.pop('AboveMedianPrice')
x = df
```

```python
from sklearn.model_selection import train_test_split as tts
x_train,x_test,y_train,y_test = tts(x,y,test_size=.2,random_state=0)
```

```python
model = Sequential()

model.add(Dense(10,input_dim=10,kernel_initializer='normal',activation='relu'))
model.add(Dense(6,input_dim=10,kernel_initializer='normal',activation='relu'))
model.add(Dense(1,input_dim=10,kernel_initializer='normal'))

model.compile(loss='mean_squared_error',optimizer='adam')
```

```python
model_history = model.fit(x_train,y_train,batch_size=10,epochs=100)
```

```python
model.summary()
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import mean_absolute_error as mae
perc = str(round(mae(y_test,y_pred),2)*100)+'%'
decimal = mae(y_test,y_pred)

if decimal>=.9:
    print(f'Model is extremely accurate. ({perc})')
elif decimal>=.6:
    print(f'Model is somewhat accurate. ({perc})')
else:
    print(f'Model is not very accurate. ({perc})')
```
